```python
# 20 In a new Code block, write a function that takes a list of strings and returns True if the list 
# contains no empty strings, or False otherwise. Include code to demonstrate correctness.
# Twentieth commit. 

def no_empty(strings):
    return all(s != "" for s in strings)

# Demonstration
print(no_empty(["Hi", "Hello"]))
print(no_empty(["Hi", ""]))


# 21 In a new Code block, write a program that prompts the user to enter integers, one at a time. The 
# input ends when the user presses Enter without typing anything. The program should display the 
# sum of the entered numbers.Twenty-first commit.

total = 0

while True:
    value = input("Enter an integer: ")
    if value == "":
        break
    total += int(value)

print("Sum:", total)

# 22 In a Code block in your Notebook, write code that prompts the user to enter a string and displays 
# whether it is a palindrome. Assume the happy path.
# Twenty-second commit. 

text = input("Enter a string: ")

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")


# 23 In a new Code block, write a function that takes a list of integers and returns the average of the 
# numbers in the list. Include code outside the function to show it works correctly.Twenty-third commit. 


def average(numbers):
    return sum(numbers) / len(numbers)

# Demonstration
print(average([2, 4, 6, 8]))


# 24 In a new Code block, write a program that prompts the user to enter strings until they press Enter 
# without typing anything. The program should display the average length of the strings.
# Twenty-fourth commit.

strings = []

while True:
    s = input("Enter a string: ")
    if s == "":
        break
    strings.append(s)

average_length = sum(len(s) for s in strings) / len(strings)
print("Average length:", average_length)



# 25 In a Code block in your Notebook, write code that prompts the user to enter an integer and 
# displays whether the number is positive, negative, or zero. Assume everything goes correctly and 
# as expected.Twenty-fifth commit.

num = int(input("Enter an integer: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```
