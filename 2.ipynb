```python
# In a new Code block, write a program that prompts the user to enter strings, one at a time. The 
# input ends when the user presses Enter without typing anything. The program should display the 
# strings in reverse order of entry.
# Third commit.

bucket = []

print("Start throwing words in! (Press Enter to stop)")

# 2. Keep doing this forever...
while True:
    # Ask for a word
    word = input("> ")
    
    # If the bucket is empty (user just pressed Enter), STOP!
    if word == "":
        break
    
    # Otherwise, toss the word into the bucket
    bucket.append(word)

# 3. Now, read the items starting from the LAST one we tossed in
print("Here is your list in reverse:")
for item in reversed(bucket):
    print(item)
```
