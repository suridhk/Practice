```python
# 12 In a new Code block, write a function that takes a list of integers and returns True if the list is 
# sorted in ascending order, or False otherwise. Include code outside the function to demonstrate 
#correctness.Twelfth commit

def is_sorted(numbers):
    return numbers == sorted(numbers)

# Demonstration
print(is_sorted([1, 2, 3, 4]))
print(is_sorted([3, 1, 4]))


# 13

strings = []

while True:
    s = input("Enter a string: ")
    if s == "":
        break
    strings.append(s)

print("Longest string:", max(strings, key=len))



# 14 In a Code block in your Notebook, write code that prompts the user to enter an integer and 
# displays whether it is a multiple of 5. Assume everything goes correctly and as expected.
# Fourteenth commit

num = int(input("Enter an integer: "))

if num % 5 == 0:
    print("The number is a multiple of 5.")
else:
    print("The number is not a multiple of 5.")


# 15 n a new Code block, write a function that takes a list of strings and returns True if all the strings 
# start with a capital letter, or False otherwise. Include some code outside the function that shows 
# that the function works correctly.
# Fifteenth commit.
 

def all_capital(strings):
    return all(s[0].isupper() for s in strings)

# Demonstration
print(all_capital(["Hello", "World"]))
print(all_capital(["Hello", "world"]))



#16 In a new Code block, write a program that prompts the user to enter strings, one at a time. The 
# input ends when the user just presses Enter. Then the program should display the strings in lower 
# case, in alphabetical order.
# Sixteenth commit.

strings = []

while True:
    s = input("Enter a string: ")
    if s == "":
        break
    strings.append(s.lower())

for s in sorted(strings):
    print(s)


# 17 In a Code block in your Notebook, write code that prompts the user to enter a string and displays 
# the number of vowels in the string. Assume the happy path.
# Seventeenth commit.


text = input("Enter a string: ")
vowels = "aeiouAEIOU"

count = 0
for char in text:
    if char in vowels:
        count += 1

print("Number of vowels:", count)



# 18 In a new Code block, write a function that takes as its parameter a list of integers. The function 
# should return True if every integer in the list is between 0 and 100 inclusive, or False otherwise. 
#Include code outside the function that shows it works correctly.Eighteenth commit. 

def valid_range(numbers):
    return all(0 <= n <= 100 for n in numbers)

# Demonstration
print(valid_range([10, 50, 100]))
print(valid_range([10, 150]))


# 19 In a Code block in your Notebook, write code that prompts the user to enter an integer 
# representing a mark on an exam. The program should display whether the mark is a pass (40 or 
# more). Assume everything goes correctly and as expected.
# Nineteenth commit.

mark = int(input("Enter exam mark: "))

if mark >= 40:
    print("Pass")
else:
    print("Fail")
```
