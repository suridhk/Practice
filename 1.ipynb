```python
#  1. In a new Code block, write a function that takes a list of integers and returns True if all numbers 
#     are unique, or False otherwise. Include code outside the function that shows it works correctly.
#     First commit.


def int(numbers):
    o = len(numbers)
    s = set(numbers)
    f = len(set(numbers))

    return o == f 

numbers= [10, 20, 30, 40]
result = int(numbers)

print(result)
```

```python
# 2 .In a Code block in your Notebook, write code that prompts the user to enter a number 
# representing the time a runner has taken to run a lap of a track. The program should display 
# whether the time is below the standard time of 50.5 seconds. Assume the happy path.
# Second commit. 

number = float(input("Enter how much time did the runner take:"))
v = 50.5

if number > v:
    print("The runner is great.")
elif number == v:
    print("The runner is average.")
else:
    print("The runner didn't reach the standard time.")

```
