```python
# In a new Code block, write a function that takes a list of integers and returns True if all numbers 
#   are unique, or False otherwise. Include code outside the function that shows it works correctly.
#   First commit.

def int(numbers):
    o = len(numbers)
    s = set(numbers)
    f = len(set(numbers))

    return o == f 

numbers= [10, 20, 30, 20]
result = int(numbers)

print(result)
```
