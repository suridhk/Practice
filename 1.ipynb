```python
#  1. In a new Code block, write a function that takes a list of integers and returns True if all numbers 
#     are unique, or False otherwise. Include code outside the function that shows it works correctly.
#     First commit.


def int(numbers):
    o = len(numbers)
    s = set(numbers)
    f = len(set(numbers))

    return o == f 

numbers= [10, 20, 30, 40]
result = int(numbers)

print(result)
```

```python
# 2 .In a Code block in your Notebook, write code that prompts the user to enter a number 
# representing the time a runner has taken to run a lap of a track. The program should display 
# whether the time is below the standard time of 50.5 seconds. Assume the happy path.
# Second commit. 

number = float(input("Enter how much time did the runner take:"))
v = 50.5

if number > v:
    print("The runner is great.")
elif number == v:
    print("The runner is average.")
else:
    print("The runner didn't reach the standard time.")

```

```python
# In a new Code block, write a program that prompts the user to enter strings, one at a time. The 
# input ends when the user presses Enter without typing anything. The program should display the 
# strings in reverse order of entry.
# Third commit.

bucket = []

print("Start throwing words in! (Press Enter to stop)")

# 2. Keep doing this forever...
while True:
    # Ask for a word
    word = input("> ")
    
    # If the bucket is empty (user just pressed Enter), STOP!
    if word == "":
        break
    
    # Otherwise, toss the word into the bucket
    bucket.append(word)

# 3. Now, read the items starting from the LAST one we tossed in
print("Here is your list in reverse:")
for item in reversed(bucket):
    print(item)
```
