```python
# 5. In a new Code block, write a function that takes a list of strings and returns True if one of the 
# strings is "coconut" (in lower case), or False otherwise. Include code outside the function that 
# shows that the function works correctly.
# Fifth commit. 

def check_for_coconut(item_list):
    # Check if the exact string "coconut" exists in the list
    if "coconut" in item_list:
        return True
    else:
        return False

# --- Testing the code ---
list_with_coconut = ["apple", "banana", "coconut", "date"]
list_without_coconut = ["car", "bus", "train"]

# This should print True
print(f"List 1 has coconut? {check_for_coconut(list_with_coconut)}")

# This should print False
print(f"List 2 has coconut? {check_for_coconut(list_without_coconut)}")
```
